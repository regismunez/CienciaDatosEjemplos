## 🎯 Introducción a CRISP-DM

# 🔧 FASE 3: PREPARACIÓN DE LOS DATOS

Limpieza y transformación:
- Renombrar columnas
- Limpiar valores
- Crear variables derivadas
- Categorizar datos


````python
# Crear copia y renombrar columnas principales
df_clean = df.copy()

columnas_renombradas = {
    'sa_01': 'genero',
    'sa_02': 'edad',
    'sa_03': 'lugar_origen',
    'sa_04': 'planea_estudiar',
    'sb_05': 'nivel_educativo_aspirado',
    'sb_11': 'ingreso_actual',
    'sb_12': 'ingreso_aspiracional'
}

df_clean = df_clean.rename(columns=columnas_renombradas)
print("✅ Columnas renombradas")
````

````python
# Limpiar valores y crear variables derivadas
df_clean['genero'] = df_clean['genero'].str.replace(r'^\d+\.\s*', '', regex=True)
df_clean['nivel_educativo_aspirado'] = df_clean['nivel_educativo_aspirado'].str.replace(r'^\d+\.\s*', '', regex=True)

df_clean['ingreso_actual'] = pd.to_numeric(df_clean['ingreso_actual'], errors='coerce')
df_clean['ingreso_aspiracional'] = pd.to_numeric(df_clean['ingreso_aspiracional'], errors='coerce')

# Variables derivadas
df_clean['brecha_ingreso'] = df_clean['ingreso_aspiracional'] - df_clean['ingreso_actual']
df_clean['ratio_aspiracion'] = (df_clean['ingreso_aspiracional'] / df_clean['ingreso_actual']).replace([np.inf, -np.inf], np.nan)

# Categorías
df_clean['grupo_edad'] = pd.cut(df_clean['edad'], bins=[0, 17, 18, 19, 100],
                                 labels=['<18', '18', '19', '>19'])

df_clean['categoria_ingreso_asp'] = pd.cut(df_clean['ingreso_aspiracional'],
                                             bins=[0, 5000, 10000, 20000, 50000, np.inf],
                                             labels=['Muy Bajo (<5K)', 'Bajo (5-10K)', 'Medio (10-20K)',
                                                    'Alto (20-50K)', 'Muy Alto (>50K)'])

print("✅ Datos limpiados y variables creadas")
````

````python
# Resumen de datos limpios
print("="*80)
print("📊 RESUMEN DE DATOS LIMPIOS")
print("="*80)
print(f"\nTotal registros: {len(df_clean):,}")
print(f"\nDistribución por Género:")
print(df_clean['genero'].value_counts())
print(f"\nNivel Educativo Aspirado:")
print(df_clean['nivel_educativo_aspirado'].value_counts())
print(f"\nEstadísticas de Ingresos:")
display(df_clean[['ingreso_actual', 'ingreso_aspiracional', 'brecha_ingreso']].describe())
````

## 📊 Tipo 1: Visualizaciones Univariadas

**Propósito:** Entender distribuciones de variables individuales

**Técnicas:** Histogramas, densidad, box plots, barras


````python
# 1A. Histograma, Densidad y Box Plot de Edad
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Histograma
axes[0].hist(df_clean['edad'].dropna(), bins=20, color='skyblue', edgecolor='black')
axes[0].set_title('Histograma: Distribución de Edad', fontweight='bold')
axes[0].set_xlabel('Edad')
axes[0].set_ylabel('Frecuencia')
axes[0].grid(alpha=0.3)

# Densidad
df_clean['edad'].dropna().plot(kind='kde', ax=axes[1], color='coral', linewidth=2)
axes[1].set_title('Densidad: Distribución de Edad', fontweight='bold')
axes[1].set_xlabel('Edad')
axes[1].grid(alpha=0.3)

# Box plot
axes[2].boxplot(df_clean['edad'].dropna())
axes[2].set_title('Box Plot: Distribución de Edad', fontweight='bold')
axes[2].set_ylabel('Edad')
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS UNIVARIADOS:")
print("- HISTOGRAMA: Muestra frecuencia exacta")
print("- DENSIDAD: Revela forma suave de distribución")
print("- BOX PLOT: Identifica outliers y cuartiles")
````

````python
# 1B. Gráficos de Barras para variables categóricas
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Género
genero_counts = df_clean['genero'].value_counts()
axes[0,0].bar(genero_counts.index, genero_counts.values, color=['#FF6B9D', '#4ECDC4'])
axes[0,0].set_title('Distribución por Género', fontweight='bold')
for i, v in enumerate(genero_counts.values):
    axes[0,0].text(i, v + 50, str(v), ha='center', fontweight='bold')

# Nivel educativo
nivel_counts = df_clean['nivel_educativo_aspirado'].value_counts()
axes[0,1].barh(nivel_counts.index, nivel_counts.values, color='steelblue')
axes[0,1].set_title('Nivel Educativo Aspirado', fontweight='bold')
axes[0,1].set_xlabel('Cantidad')

# Lugar origen (top 10)
lugar_counts = df_clean['lugar_origen'].value_counts().head(10)
axes[1,0].bar(range(len(lugar_counts)), lugar_counts.values, color='coral')
axes[1,0].set_title('Top 10 Lugares de Origen', fontweight='bold')
axes[1,0].set_xticks(range(len(lugar_counts)))
axes[1,0].set_xticklabels(lugar_counts.index, rotation=45, ha='right')

# Grupo edad
edad_counts = df_clean['grupo_edad'].value_counts().sort_index()
axes[1,1].bar(edad_counts.index, edad_counts.values, color='mediumpurple')
axes[1,1].set_title('Distribución por Grupo de Edad', fontweight='bold')

plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS:")
print("- BARRAS VERTICALES: Fácil comparación de magnitudes")
print("- BARRAS HORIZONTALES: Mejores para etiquetas largas")
````

## 📊 Tipo 2: Visualizaciones Bivariadas

**Propósito:** Explorar relaciones entre dos variables

**Técnicas:** Scatter plots, barras agrupadas, líneas


````python
# 2A. Scatter Plot: Ingreso Actual vs Aspiracional
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Scatter básico con color por edad
scatter = axes[0].scatter(df_clean['ingreso_actual'], df_clean['ingreso_aspiracional'],
                          c=df_clean['edad'], cmap='viridis', alpha=0.6, s=50)
axes[0].plot([0, df_clean['ingreso_actual'].max()], [0, df_clean['ingreso_actual'].max()],
             'r--', label='Línea de igualdad')
axes[0].set_xlabel('Ingreso Actual (Bs)')
axes[0].set_ylabel('Ingreso Aspiracional (Bs)')
axes[0].set_title('Relación: Ingreso Actual vs Aspiracional\n(color = edad)', fontweight='bold')
axes[0].legend()
axes[0].grid(alpha=0.3)
plt.colorbar(scatter, ax=axes[0], label='Edad')

# Scatter por género
for genero in df_clean['genero'].unique():
    data = df_clean[df_clean['genero'] == genero]
    axes[1].scatter(data['ingreso_actual'], data['ingreso_aspiracional'],
                    label=genero, alpha=0.6, s=50)
axes[1].plot([0, df_clean['ingreso_actual'].max()], [0, df_clean['ingreso_actual'].max()],
             'k--', alpha=0.3)
axes[1].set_xlabel('Ingreso Actual (Bs)')
axes[1].set_ylabel('Ingreso Aspiracional (Bs)')
axes[1].set_title('Relación por Género', fontweight='bold')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS DE SCATTER PLOTS:")
print("- Revelan CORRELACIONES y patrones no lineales")
print("- El COLOR añade una tercera dimensión")
print("- LÍNEAS DE REFERENCIA contextualizan los datos")
````

````python
# 2B. Barras Agrupadas
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Nivel educativo por género
nivel_genero = pd.crosstab(df_clean['nivel_educativo_aspirado'], df_clean['genero'])
nivel_genero.plot(kind='bar', ax=axes[0], color=['#FF6B9D', '#4ECDC4'])
axes[0].set_title('Nivel Educativo por Género', fontweight='bold')
axes[0].set_ylabel('Cantidad')
axes[0].tick_params(axis='x', rotation=45)
axes[0].legend(title='Género')

# Ingreso promedio por nivel y género
ingreso_pivot = df_clean.pivot_table(values='ingreso_aspiracional',
                                       index='nivel_educativo_aspirado',
                                       columns='genero', aggfunc='mean')
ingreso_pivot.plot(kind='barh', ax=axes[1], color=['#FF6B9D', '#4ECDC4'])
axes[1].set_title('Ingreso Aspiracional Promedio\npor Nivel y Género', fontweight='bold')
axes[1].set_xlabel('Ingreso Promedio (Bs)')
axes[1].legend(title='Género')

plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS BARRAS AGRUPADAS:")
print("- Comparaciones DIRECTAS entre categorías")
print("- Identifican BRECHAS entre grupos")
````

## 📊 Tipo 3: Visualizaciones Multivariadas

**Propósito:** Analizar relaciones complejas entre múltiples variables

**Técnicas:** Burbujas, heatmaps, pair plots


````python
# 3A. Gráfico de Burbujas (4 dimensiones)
df_bubble = df_clean[['edad', 'ingreso_actual', 'ingreso_aspiracional',
                       'brecha_ingreso', 'genero']].dropna()

fig, ax = plt.subplots(figsize=(14, 8))

for genero in df_bubble['genero'].unique():
    data = df_bubble[df_bubble['genero'] == genero]
    ax.scatter(data['edad'], data['ingreso_aspiracional'],
               s=data['brecha_ingreso']/50, alpha=0.5, label=genero)

ax.set_xlabel('Edad')
ax.set_ylabel('Ingreso Aspiracional (Bs)')
ax.set_title('Burbujas: Edad vs Ingreso\n(Tamaño = Brecha de Ingreso)', fontweight='bold')
ax.legend(title='Género')
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS BURBUJAS:")
print("- Visualizan 4 DIMENSIONES simultáneamente:")
print("  • X: Edad")
print("  • Y: Ingreso Aspiracional")
print("  • Tamaño: Brecha")
print("  • Color: Género")
````

````python
# 3B. Heatmap de Correlación
numeric_cols = df_clean.select_dtypes(include=[np.number]).columns
correlation = df_clean[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=1)
plt.title('Heatmap: Correlaciones entre Variables Numéricas', fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS HEATMAPS:")
print("- Revelan RELACIONES entre todas las variables")
print("- Rojo = correlación positiva fuerte")
print("- Azul = correlación negativa fuerte")
````

## 📊 Tipo 4: Visualizaciones de Composición

**Propósito:** Mostrar partes del todo

**Técnicas:** Pastel, dona, barras apiladas


````python
# 4A. Gráficos de Pastel y Dona
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Pastel - Género
genero_counts = df_clean['genero'].value_counts()
axes[0].pie(genero_counts.values, labels=genero_counts.index, autopct='%1.1f%%',
            colors=['#FF6B9D', '#4ECDC4'], startangle=90)
axes[0].set_title('Composición por Género', fontweight='bold')

# Dona - Nivel educativo
nivel_counts = df_clean['nivel_educativo_aspirado'].value_counts()
wedges, texts, autotexts = axes[1].pie(nivel_counts.values, labels=nivel_counts.index,
                                        autopct='%1.1f%%', startangle=90, pctdistance=0.85)
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
axes[1].add_artist(centre_circle)
axes[1].set_title('Nivel Educativo (Dona)', fontweight='bold')

# Pastel - Grupo edad
edad_counts = df_clean['grupo_edad'].value_counts()
axes[2].pie(edad_counts.values, labels=edad_counts.index, autopct='%1.1f%%', startangle=90)
axes[2].set_title('Composición por Edad', fontweight='bold')

plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS PASTEL/DONA:")
print("- Excelentes para PROPORCIONES simples")
print("- Gráficos de DONA son más modernos")
print("- LIMITACIÓN: Difíciles con >5-6 categorías")
````

````python
# 4B. Barras Apiladas
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Apiladas porcentuales
nivel_genero_pct = pd.crosstab(df_clean['nivel_educativo_aspirado'],
                                df_clean['genero'], normalize='index') * 100
nivel_genero_pct.plot(kind='bar', stacked=True, ax=axes[0], color=['#FF6B9D', '#4ECDC4'])
axes[0].set_title('Distribución % de Género\npor Nivel Educativo', fontweight='bold')
axes[0].set_ylabel('Porcentaje (%)')
axes[0].tick_params(axis='x', rotation=45)

# Apiladas absolutas
ingreso_genero = pd.crosstab(df_clean['categoria_ingreso_asp'], df_clean['genero'])
ingreso_genero.plot(kind='barh', stacked=True, ax=axes[1], color=['#FF6B9D', '#4ECDC4'])
axes[1].set_title('Distribución de Género\npor Categoría de Ingreso', fontweight='bold')
axes[1].set_xlabel('Cantidad')

plt.tight_layout()
plt.show()

print("\n📌 INSIGHTS BARRAS APILADAS:")
print("- Muestran TOTALES y COMPOSICIÓN")
print("- Apiladas 100%: comparar proporciones")
print("- Apiladas absolutas: magnitudes reales")
````

## 📊 Tipo 5: Visualizaciones Interactivas (Plotly)

**Propósito:** Exploración dinámica

**Ventajas:** Zoom, tooltips, filtrado dinámico


````python
# 5A. Scatter Interactivo
fig = px.scatter(df_clean.dropna(subset=['ingreso_actual', 'ingreso_aspiracional']),
                 x='ingreso_actual', y='ingreso_aspiracional',
                 color='genero', size='brecha_ingreso',
                 hover_data=['edad', 'nivel_educativo_aspirado'],
                 title='Exploración Interactiva: Ingreso Actual vs Aspiracional',
                 labels={'ingreso_actual': 'Ingreso Actual (Bs)',
                        'ingreso_aspiracional': 'Ingreso Aspiracional (Bs)',
                        'genero': 'Género', 'brecha_ingreso': 'Brecha'},
                 color_discrete_map={'MUJER': '#FF6B9D', 'HOMBRE': '#4ECDC4'})
fig.update_layout(height=600)
fig.show()

print("\n📌 INSIGHTS INTERACTIVOS:")
print("- TOOLTIPS: información detallada al pasar el mouse")
print("- ZOOM/PAN: análisis de regiones específicas")
print("- Ideales para PRESENTACIONES y DASHBOARDS")
````

````python
# 5B. Gráfico Sunburst (Jerarquía)
df_sunburst = df_clean.groupby(['genero', 'nivel_educativo_aspirado', 'grupo_edad']).size().reset_index(name='count')

fig = px.sunburst(df_sunburst,
                  path=['genero', 'nivel_educativo_aspirado', 'grupo_edad'],
                  values='count',
                  title='Jerarquía: Género → Nivel Educativo → Edad',
                  color='count', color_continuous_scale='Viridis')
fig.update_layout(height=700)
fig.show()

print("\n📌 INSIGHTS SUNBURST:")
print("- Excelentes para DATOS JERÁRQUICOS")
print("- DRILL-DOWN interactivo")
print("- Muestran PROPORCIONES y JERARQUÍAS simultáneamente")
````

## Parte 1: Insights por Tipo de Visualización

### 1.1 Visualizaciones Univariadas
**Insights más evidentes:**
- **Histogramas:** Forma de distribución (normal, sesgada, bimodal)
- **Box Plots:** Outliers y asimetría instantánea
- **Densidad:** Distribuciones continuas suaves

**Ejemplo:** Los histogramas de edad mostraron concentración en 17-19 años, mientras box plots revelaron estudiantes atípicos mayores.

### 1.2 Visualizaciones Bivariadas
**Insights más evidentes:**
- **Scatter Plots:** Correlaciones, tendencias, agrupaciones
- **Barras Agrupadas:** Diferencias entre grupos

**Ejemplo:** Scatter plots revelaron que todos aspiran a ingresos mayores, independientemente del punto de partida.

### 1.3 Visualizaciones Multivariadas
**Insights más evidentes:**
- **Burbujas:** 4 dimensiones simultáneas
- **Heatmaps:** Relaciones entre todas las variables
- **Pair Plots:** Múltiples patrones bivariados

**Ejemplo:** Heatmap mostró correlación moderada entre ingresos, pero relación más compleja con brecha.

### 1.4 Visualizaciones de Composición
**Insights más evidentes:**
- **Pastel/Dona:** Proporciones relativas
- **Barras Apiladas:** Evolución de composición

**Ejemplo:** Pastel mostró balance de género, barras apiladas revelaron diferencias en aspiraciones.

### 1.5 Visualizaciones Interactivas
**Insights más evidentes:**
- **Scatter Interactivos:** Detalles de puntos específicos
- **Sunburst:** Drill-down en jerarquías

**Ejemplo:** Scatter interactivo permitió explorar outliers específicos con tooltips.


## Parte 2: Cómo Diferentes Técnicas Facilitan Diferentes Descubrimientos

### 2.1 Principios de Percepción Visual

**Jerarquía de Precisión (Cleveland & McGill, 1984):**
1. Posición en escala común (MÁS PRECISO) - scatter, barras
2. Posición en escalas no alineadas
3. Longitud, dirección, ángulo - barras, líneas
4. Área - burbujas
5. Volumen, curvatura
6. Sombreado, color (MENOS PRECISO) - heatmaps

### 2.2 Matriz de Efectividad

La efectividad depende del tipo de pregunta:
- **Distribución:** Histograma > Densidad > Box Plot
- **Outliers:** Box Plot > Scatter > Histograma
- **Categoría mayor:** Barras > Pastel > Scatter
- **Correlación:** Scatter > Heatmap > Líneas
- **Proporciones:** Pastel > Barras Apiladas > Scatter
- **Clusters:** Scatter > Coordenadas Paralelas > Heatmap

### 2.3 Factores que Influyen en Efectividad

**A. Complejidad de Datos**
- Simple (1-2 vars): Barras, pastel, histogramas
- Moderado (2-4 vars): Scatter, burbujas, barras agrupadas
- Complejo (5+ vars): Parallel coordinates, heatmaps

**B. Audiencia**
- Público general: Barras, pastel, líneas
- Analistas: Heatmaps, coordenadas paralelas
- Ejecutivos: Dashboards interactivos

**C. Objetivo**
- Exploración: Pair plots, scatter interactivos
- Confirmación: Gráficos específicos
- Presentación: Visualizaciones pulidas y simples

**D. Medio**
- Papel/PDF: Estáticas de alta calidad
- Digital: Interactivas
- Dashboard: Actualizables y filtrados

# 🎯 CONCLUSIONES Y RECOMENDACIONES

## Conclusiones Principales

### 1. No Existe una "Mejor" Visualización Universal
Cada tipo tiene fortalezas específicas. La efectividad depende del contexto, audiencia y objetivo.

### 2. Hallazgos del Dataset de Estudiantes

**Demográficos:**
- Edad concentrada en 17-19 años (histogramas)
- Género balanceado (pastel)
- Algunos estudiantes atípicos mayores (box plots)

**Aspiraciones:**
- Mayoría aspira a educación universitaria (barras)
- Fuerte correlación ingreso actual-aspiracional (scatter)
- Aspiraciones de ingreso bimodales (violin plots)
- Diferencias sutiles por género (barras agrupadas)

**Económicos:**
- Brecha de ingreso consistentemente alta (burbujas)
- Outliers evidentes en grupos específicos (scatter interactivos)
- Relaciones complejas variables económicas-demográficas (coordenadas paralelas)

### 3. Tabla de Insights Únicos

| Tipo | Insight Único |
|------|---------------|
| Histogramas | Forma exacta de distribuciones |
| Box Plots | Identificación precisa de outliers |
| Scatter | Correlaciones no lineales |
| Burbujas | Interacciones 4D |

## Recomendaciones

### Workflow Recomendado

**FASE 1: EXPLORACIÓN INICIAL**
- Histogramas para cada variable numérica
- Barras para cada variable categórica
- Box plots para identificar outliers
- Pair plot para vista panorámica

**FASE 2: ANÁLISIS PROFUNDO**
- Scatter plots para correlaciones específicas
- Violin plots para distribuciones complejas
- Heatmaps para relaciones múltiples
- Coordenadas paralelas para patrones multivariados

**FASE 3: COMUNICACIÓN**
- Seleccionar 3-5 visualizaciones clave
- Pulir y simplificar
- Añadir anotaciones y contexto
- Crear versiones interactivas si aplica

## Reflexión sobre CRISP-DM y Visualización

CRISP-DM proporciona marco estructurado para:
- **Entender** el contexto antes de visualizar
- **Explorar** datos sistemáticamente
- **Preparar** datos apropiadamente
- **Modelar** (visualizar) de múltiples formas
- **Evaluar** efectividad
- **Desplegar** las más efectivas

Este enfoque asegura:
1. No quedarnos con la primera visualización
2. Explorar múltiples perspectivas
